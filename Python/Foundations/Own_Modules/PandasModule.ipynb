```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Karthik", "Meena", "Rahul"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"],
    "Attendance": [82, 91, 76, 68, 88],
    "Python": [85, 92, 76, 65, 88],
    "SQL": [78, 88, 82, 70, 91],
    "Excel": [90, 95, 80, 72, 86]
}
table = pd.DataFrame(data)
print(data)

print(table)

print("First 2 Rows:\n",table.head(2))
print("Last 2 Rows:\n",table.tail(2))

print("Random 3 Rows:\n",table.sample(2))

print("No Of (Rows,Column): ",table.shape)

print("Header:\n",table.columns)

print("Info:\n",table.info())

table.describe()
```

```python
mean = table[["Python","SQL","Excel"]].mean()
median = table[["Python","SQL","Excel"]].median()
mode = table[["Python","SQL","Excel"]].mode()

print(mean)
print(median)
print(mode)  
```
