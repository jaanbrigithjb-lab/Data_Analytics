```python
import numpy as np
m = [10,20,30,40,50]
marks = np.array(m)
marks = marks + 5
grade = marks>=45

s = marks.size

for i in range(s):
    print(marks[i]>=45)
    if marks[i]>=45 == True:
        print("Pass")
    else:
        print("Fail")

print(marks)
print(grade)
```
