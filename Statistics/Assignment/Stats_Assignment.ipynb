```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("Dataset.xlsx", sheet_name="Dataset")

np.random.seed(42)

sample_10 = df.sample(n=10, replace=False)

print(sample_10)
```

```python
sample_mean = sample_10["Order_Value"].mean()

print("Sample Mean:", sample_mean)
```

```python
np.random.seed(42)

sample_means = []

for i in range(100):
    sample = df.sample(n=10, replace=False)
    sample_mean = sample["Order_Value"].mean()
    sample_means.append(sample_mean)

print(sample_means)
```

```python
np.random.seed(42)

sample_means = []

for i in range(100):
    sample = df.sample(n=10, replace=False)
    mean_value = sample["Order_Value"].mean()
    sample_means.append(mean_value)

print("Number of sample means:", sample_means)
```

```python
plt.hist(sample_means, bins=10)

plt.xlabel("Sample Mean of Order Value")
plt.ylabel("Frequency")
plt.title("Distribution of 100 Sample Means")

plt.show()
```

```python
plt.hist(sample_means, bins=10)

plt.xlabel("Sample Mean of Order Value")
plt.ylabel("Frequency")
plt.title("Distribution of 100 Sample Means")

plt.show()
```

```python
print(len(set(sample_means)) == 1)
```

```python
print("Each sample contains different randomly selected customers,")
print("so the Order_Value values differ and therefore the sample means differ.")
```

```python
np.random.seed(42)

sample_means_25 = []

for i in range(100):
    sample = df.sample(n=25, replace=False)
    mean_value = sample["Order_Value"].mean()
    sample_means_25.append(mean_value)

plt.hist(sample_means_25, bins=10)

plt.xlabel("Sample Mean of Order Value")
plt.ylabel("Frequency")
plt.title("Distribution of 100 Sample Means (Sample Size = 25)")

plt.show()
```

```python
print("Sample size 25 generally produces more stable sample means than sample size 10.")
```
